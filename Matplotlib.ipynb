```python
x=data['Country Code']
y=data['Birth rate']
plt.plot(x,y,marker='*',label='cc vs br',color='k')
plt.xlabel('Country Code')
plt.ylabel('Birth rate')
plt.title('line plot')
plt.grid(True)
plt.legend()
plt.show()
```

```python
x=data['Country Code']
y=data['Birth rate']
plt.plot(x,y,'*-.m',label='cc vs br')
plt.xlabel('Country Code')
plt.ylabel('Birth rate')
plt.title('line plot')
plt.grid(True)
plt.legend()
plt.show()
```

```python
x=data['Country Code']
y=data['Birth rate']
plt.scatter(x,y,marker='*',label='cc vs br',color='k')
plt.xlabel('Country Code')
plt.ylabel('Birth rate')
plt.title('line plot')
plt.grid(True)
plt.legend()
plt.show()
```

```python
font1={'family':'serif','color':'cyan','size':10}
font2={'family':'serif','color':'green','size':7}
plt.scatter(x,y,label='cc vs br')
plt.xlabel('Country Code',fontdict=font2)
plt.ylabel('Birth rate',fontdict=font2)
plt.title('line plot',fontdict=font1,loc='center')
plt.grid(True)
plt.legend()
plt.show()
```

```python
font1={'family':'serif','color':'k','size':20}
font2={'family':'serif','color':'r','size':15}
plt.bar(x,y,label='cc vs br',color='b')
plt.xlabel('Country Code',fontdict=font2)
plt.ylabel('Birth rate',fontdict=font2)
plt.title('line plot',fontdict=font1)
plt.grid(True)
plt.legend()
plt.show()
```
